```python
#install the libraries:-
import pandas as pd
import numpy as np
```

```python
#Task1/
#read the file
df = pd.read_csv("/content/sales_data_sample.csv", encoding="latin1")
print(f"first 5 rows is: {df.head(5)}")
print(f"last 5 rows is: {df.tail(5)}")
```

```python
#Task2/
print(f"Sum of rows = {df.shape[0]}")
print(f"Sum of columns = {df.shape[1]}")
print(f"The name of the columns is: {list(df.columns)}")
print(f"The Types of the columns is: {df.dtypes}")
```

```python
#Task3/
display(df.describe())
display(df.describe(include="object"))
```

```python
#Task4/
print(np.mean(df["SALES"]))
print(np.median(df["SALES"]))
print(np.min(df["SALES"]))
print(np.max(df["SALES"]))
```

```python
#Task5/
display(df[["ORDERLINENUMBER","CUSTOMERNAME","SALES"]])
```

```python
#Task6/
display(df.iloc[10:21:])
```

```python
#Task7/
display(df.loc[df["ORDERNUMBER"]==10100])
```

```python
#Task8/

sales_df = pd.DataFrame({
    "PRODUCTLINE" : df['PRODUCTLINE'],
"COUNTRY" : df["COUNTRY"],
"SALES" :df["SALES"]
})
display(sales_df)
```

```python
#Task9/
display(df.loc[df['SALES']>5000])
```

```python
#Task10/
display(df[df['COUNTRY']=="USA"])
```

```python
#Task11/
display(df[df["COUNTRY"].isin(["USA","France"])])
```

```python
#Task12/
display(df[ (df["PRODUCTLINE"] == 'Classic Cars')
&
            (df["SALES"] > 3000)
            ])
```

```python
#Task14
df['TOTAL_VALUE']= df['QUANTITYORDERED'] * df['PRICEEACH']
display(df['TOTAL_VALUE'])
```

```python
#Task15
df['DISCOUNT_FLAG'] = np.where(df["SALES"] > 5000, "High", "Normal")
df['DISCOUNT_FLAG']
```

```python
#Task17
df['BIG_ORDER'] = np.where(df["QUANTITYORDERED"] >= 40, "Yes", "No")
df['BIG_ORDER']
```

```python
#Task18
display(df.isnull().sum())
print("---------")
display(df.apply(lambda x: x.isnull().sum() / len(df) * 100))
```

```python
#Task19
duplicate = df.duplicated().sum()
print(f"Number of duplicate rows: {duplicate}")
```

```python
#Task21
print(f"the sum of each sale {df.groupby('COUNTRY')['SALES'].sum().sort_values(ascending=False)} ")
```

```python
#Task22
print(df.groupby('PRODUCTLINE')['SALES'].mean())
```

```python
#Task23
display(df.groupby('COUNTRY')['SALES'].agg(['sum','count','max','mean']))
```

```python
#Task24
sales_by_country_product = df.pivot_table(
    index='COUNTRY',
    columns='PRODUCTLINE',
    values='SALES',
    aggfunc='sum',
    fill_value=0

)
display(sales_by_country_product)
```

```python
#Task25
orders_by_year_status = df.pivot_table(
    index='YEAR_ID',
    columns='STATUS',
    values='SALES',
    aggfunc='count',
    fill_value=0

)
display(orders_by_year_status)
```

```python
#Task 27
df['ORDER-DATE'] = pd.to_datetime(df['ORDERDATE'])
```

```python
#Task28
df['year'] = df['ORDER-DATE'].dt.year
df['month'] = df['ORDER-DATE'].dt.month
df['day'] = df['ORDER-DATE'].dt.day
df['day_name'] = df['ORDER-DATE'].dt.day_name()

display(df['year'])
display(df['month'])
display(df['day'])
display(df['day_name'])
```
